### Caderno de Conceitos Estatísticos Essenciais com Statsmodels

**1. O que é análise estatística?**

Análise estatística e a ciência e a prática de coletar, explorar, interpretar e apresentar dados para descobrir padrões, tendências e relações entre variáveis. Ela utiliza modelos matemáticos e estatísticos para transformar dados brutos em insights compreensíveis.

O objeto principal da análise estatística não é apenas descrever o que aconteceu no passado (estatística descritiva), mas também usar esses dados para tirar conclusões sobre o mundo em geral, muitas vezes aplicando a teoria da probabilidade para quantificar a incerteza dessas conclusões (estatística inferencial).

Análise estatística fornece as ferramentas para:

 - Resumir dados: Através de médias, medianas, desvio-padrão, etc.

 - Visualizar dados: usando gráficos para identificar padrões.

 - Testar hipóteses: Para tomar decisões baseadas em evidências (ex: "O novo design do site realmente aumentou as vendas?")

 - Qualificar incereza: Utilizando intervalos de confiança e p-valores para entender a confiabilidade dos resultados.

**2.Principais tipos de análise estatística**

A análise estatística é geralmente dividida em duas grandes categorias:

*1. Estatística Descritiva*

O objetivo e resumir e organizar os dados de um estudo ou amostra. Não tenta tirar conclusões para além dos dados analisados. As principais ferramentas incluem:

 - Medidas de tendência Central: Como a média, a mediana e a moda, que representam um valor típico do conjunto de dados.

 - Medidas de dispensão: Como o desvio-padrão, a variância e a amplitude, que indicam o quão espalhados os dados estão.

 - Visualizações Gráficas: Gráficos de barras, histogramas, e diagramas de dispersão que ajudam a visualizar os dados.

*2. Estatística inferencial*

O seu objetivo é usar os dados de uma amostra para fazer generalizações, previsões ou inferências sobre uma polução inteira da qual a amostra foi retirada. As técnicas mais comuns incluem:

 - Testes de hipóteses: Procedimentos formais para aceitar ou rejeitar uma hipótese estatísca (por exemplo, teste se um novo medicamento e mais eficaz que o antigo.

 - Análise de regressão: Modela a relação entre uma variável dependênte e uma ou mais variáveis independentes, permitindo faze previsões.

 - Intervalos de confiança: Fornecem um intervalo de valores que provavelmente contém o verdadeiro valor de um parâmetro da população.


**3. Diferença Entre Modelagem Estatística vs. Modelagem Preditiva**

A modelagem estatística e a modelagem preditíva compartilham muitas bases matemáticas, mas têm objetivos e filosofias diferentes. Enquanto a estatísca busca os dados e a relação entre variáveis, a modelagem preditiva tem como foco prever resultados futuros com o máximo de precisão possível, mesmo que a interpretação do modelo seja limitada.

Na modelagem estatística, o objetivo principal e explicar fenômenos, Ela busca compreender como e porque as variáveis estão relacionadas. Por exemplo, em uma regressão linear clássica, o foco é estimar o impacto de uma variável independente sobre a variável dependente, interpretando coeficientes, intervalos de confiança e significância estatística (p-valores). Aqui, explicabilidade é mais importante que a precisão. É uma abordagem comum em pesquisar científicas, economia e ciências sociais.

Já a modelagem preditiva é orientada à performance e generalização. O interesse não está em necessariamente entender  o "porquê", mas em fazer boas previsões, seja para identificar clientes propensos ao churn, prever fraudes ou recomendar produtos. Ela usa algarítimos mais complexos (como arvore de decisão, random forests, gradient boosting, redes neurais) e se apoia em técnicas de validação e ajuste fino (como cross - validation e otimização de hiperparâmetros). Aqui, a prioridade é o acervo, mesmo que o modelo seja uma "caixa-preta".

Imagine que você quer entender o causa o preço de uma casa:

 - Na modelagem estatística nós contruiriamos um modelo de regressão para explcar o preço. O modelo ajudaria a explicar : "Para cada metro quadrado, o preço aumenta, em média, em 12.000 reais, e este efeito é estatísticamente significativo (p-valor < 0.05). A presença de uma piscina aumenta o preço em 15.000 reais. " O objetivo é entender os fatores e sua importância.

 - Na modelagem preditiva testatíamos vários algorítmos para prever o preço de uma casa nova que entra no mercado. Aqui poderíamos usar um modelo complexo (como XGBoost) e dizer: "Para esta casa específica com estas 150 características, podemos prever um preço de 150.000 com um margem de erro de 3%". O modelo não sabe exatamente como explicar como cada uma das 150 características contribuiu, mas o mas o modelo é extremamente preciso em suas previsões.

A modelagem estatística explica o passado, enquanto a modelagem preditiva antecipa o futuro. A primeira busca compreender as relações e testar hipóteses, a segunda busca  construir sistemas capazes por prever com confiança e, muitas vezes, em tempo real. O profissional de dados precisa dominar ambas: entender os fundamentos estatísticos para interpretar corretamente os resultados e aplicar técnicas preditivas para transformar dados em ação.

*Use modelagem estatística quando precisar de explicação e compreensão. Use modelagem preditiva quando precisar de previsão precisas e acionáveis. As duas abordagem são ferramentas poderosas e complementares no arsenal de um cientista de dados.*

**4. Estatística Descritiva**

**Definição**: É o processo de usar números para resumir e descrever as principais características de um conjunto de dados. Em vez de olhar para centenas de dados  brutos, usamos medidas como a média (valor central), mediana (valor do meio) e desvio padrão (o quão espalhados os dados estão).

**4.1. Medidas de Tendância Central**

As medidas de tendência central indicam onde os dados se concentram, ou seja, o valor que representa o "centro" ou o "típico" do conjunto. As principais são:

 - Média: (aritimética): É a somas de todos os valores dividida pelo número de observações. Representa o valor médio do grupo, mas é sensível a outliers (valores muito altos ou baixos). Exemplo: a média [2, 3, 5] é (2+3+5)/3 = 3,33.

 - Mediana: É o valor central quando os dados são ordenados. Divide o conjunto ao meio, metade dos valores abaixo e metade acima. É menos afetada por outliers, sendo útil em distribuições assímétricas. Exemplo: em [2, 3, 100], a mediana é 3.

 - Moda: É o valor que mais se repete no conjunto. Pode haver uma (unimodal), várias (multimodal) ou nenhuma moda (quando todos os valores são únicos). Exemplo: em [2,2,3,5], a moda é 2.

Essas medidas respondem á pergunta: "Qual é o valor típico ou central dos meus dados?"

**4.2. Medidas de Dispersão**

As medidas de dispersão mostram quanto os dados se afastam do centro, ou seja, o grau de variabilidade entre as observações. As principais são:

 - Amplitude (ou intervalo): É a diferença entre o maior e menor valor. Mede a variação total, mas depende fortemente de outliers. Exemplo: se os dados vão de 10 a 40 a amplitude é 30.

 - Variância: Mede quanto, em média, os valores se afastam da média, elevando essas diferenças ao quadrado. É expressa em unidades ao quadrado. Exemplo: variância alta => dados muito dispersos.

 - Desvio padrão: É a raiz quadrada da variância, voltando à unidade original dos dados. Representa o desvio médio em relação à média. Exemplo: se o desvio padrão é 5,  os dados tendem a variar +5 ou -5 unidade em torno da média.

 - Coeficiente de variação (CV):É o desvio padrão dividido pela média, em percentual. Mostra a dispersão relativa - quanto maior o CV, mais instável é o conjuto de dados. Exemplo: CV = (desvio-padrão / média) * 100.

Resumindo

 - As média de tendência central indicam onde os dados se concentram.

 - As medidas de dispensão indicam o quanto eles variam em torno do centro.

Juntas, elas fornecem uma visão completa da distribuição e da consistência dos dados, base essencial para qualquer análise estatística ou modelo de Machine Learning.

**4.3. Análise de Correlação**

*Definição*: Correlação é uma medida estatística que expressa o grau em que dusa variáveis se movem em relação à outra.

 - *Correlação Positiva*: Quando uma variável aumenta, a outra tende a aumentar (ex: altura e peso).
 - *Correlação Negativa*: Quando uma variável aumenta, a outra tende a diminuir (ex:horas de estudo e horas de videogame).
 - *Correlação Próxima de Zero: Não há uma relação linear clara entre as variáveis.

O coeficiente de correlação varia de -1 a +1.

A correlação é, em essência, uma medida da estatística descritiva, pois descreve a força e a direção de uma relação dentro de um conjuto de dados observado.

O coeficiente de correlação (como o de Pearson ou Spearman) é calculado diretamente a partir dos dados de amostra e resume essa relação numéricamente. Por exemplo, um valor próximo de +1 indica uma relação positiva forte, enquanto um valor próximo de -1 indica uma relação negativa forte.

**5. Estatística Inferencial**

*Definição*: Estatística inferencial é o ramo da estatística que permite tirar conclusões ou fazer referências sobre uma população com base em uma amostra de dados. Em vez de apenas descrever os dados observados (como na estatística descritiva), a estatística inferencial utiliza métodos probabilisticos para estimar parâmetros populacionais, testar hipóteses e avaliar a significância de resultados.

A Estatística inferencial busca responder pergutas como:

"O que podemos inferir sobre a população, a partir da amostra?"

Ferramentas típicas incluem intervalos de confiança, testes de hipóteses, análises de variância (ANOVA) e regressão, sempre com o objetivo de transformar dados amostrais em conhecimento generalizavel e confiável.

**5.1. Teste de Hipótese (Teste t)**

*Definição*: É um método formal para tomar uma decisão sobre uma afirmação (hipótese) usando dados. O Teste t, especificamente, é usado para comparar as médias de dois grupos e determinar se a diferença entre elas é significativa ou se provavelmente ocorreu por acaso.

 - Hipótese Nula (H°): Afirmação padrão, geralmente de que "não há diferença". (Ex: "A média de altura dos dois grupos é a mesma").

 - Hipótese Alternativa (H¹ ou Ha): É a afirmação contrátia a hipótese nula (H°), ela comprova o que queremos comprovar ou evidenciar com os dados.

 - Valor-p (p-value): A probabilidade de observar os dados que temos se a hipótese nula for verdadeira. Um valor-p baixo (geralmente < 0,05) nos dá evidências para rejeitar a hipótese nula.

```python
# Vamos usar os mesmos dados de altura dos grupos A e B
# Hipotese Nula (H0): Não há diferença significativa entre as médias de altura do grupo A e do grupo B.

# Hipótese (H0): média grupo A == média grupo B => Não há diferença entre as médias dos grupos.
# Hipótese (H1): média grupo A != média grupo B => Há diferença entre as médias dos grupos.

# Realizando o Teste t com Statsmodels
# ttest_ind retorna: (estatística t, valor-p, graus de liberdade)
t_stat, p_value, df = ttest_ind(grupo_a, grupo_b)

print(f"\n--- Resultado do Test t ---\n")
print(f"Estatística t: {t_stat:.4f}")
print(f"Valor-p: {p_value:.4f}")

# Interpretando o resultado
alpha = 0.05
if p_value < alpha:
    print(f"\nConclusão: Como o valor-p ({p_value:.4f}) é menor que {alpha}, rejeitamos a hipótese nula.")
    print("Há uma diferença estatisticamente significativa entre as médias dos dois grupos.\n")
else:
    print(f"\nConclusão: Como o valor-p ({p_value:.4f}) é maior que {alpha}, não temos evidências para rejeitar a hipótese nula.")
    print("Não há diferenã estatisticamente significativa entre as médias dos dois grupos.\n")
```

**5.2. Regressão Linear Simples**

*Definição*: É uma técnica  para modelar a relação entre uma variável independênte (preditora, X) e uma variável dependente (resposta, Y) através de uma linha reta. O objetivo é não apenas prever Y a partir de X, mas também aprender a força e a direção dessa relação.

A equação é: Y = ß0+ß1x+e

 - ß0: Intercepto (onde a  linha cruza o eixo Y)
 - ß1: Coeficiente angular (quando Y muda para cada unidade de almento em X)

```python
# 1. Geração de dados de exemplos
# Suponha que estamos estudando a relação entre horas de estudo e nota em um exame
np.random.seed(42)
horas_estudo = np.random.uniform(1, 10, 50) # variável independente (X)
nota = 5 + 3 * horas_estudo + np.random.normal(0, 3, 50) # variável dependente (Y)
```

```python
# Cria o DataFrame
df_score = pd.DataFrame({'Horas_Estudo': horas_estudo, 'Nota': nota})
```

```python
df_score.head()
```

```python
# 2. Visualização dos dados
sns.scatterplot(data = df_score, x = 'Horas_Estudo', y = 'Nota')
plt.title('Relação Entre Horas de Estudo e Notas')
plt.xlabel('Horas de Estudo')
plt.ylabel('Nota')
plt.show()
```

```python
# 3. Ajuste do modelo de regressão linear
X = sm.add_constant(df_score['Horas_Estudo']) # Adiciona o intercepto, o que é requerido pelo Statsmodels
Y = df_score['Nota']
```

```python
# OLS = Ordinary Least Squares (Mínimos Quadrados Ordinários)
modelo = sm.OLS(Y, X).fit()
```

```python
# 4. Resultados do modelo
print(modelo.summary())
```

Os erros padrão (Standard Erros) apresentados assumem que a matriz de covariância dos erros está corretamente especificada.

Em outras palavras: O modelo de regressão está supondo que os erros (resíduos) seguem as condições ideais do modelo clássico de mínimos quadrados (OLS). Isso significa que o Statsmodels calculou os erros padrão, os valores-t e os p-valores sob as seguintes suposições:

 - Homoscedasticidade: a variância dos erros é constante ao longo das observações (não há heterocedasticidade).

 - Independência dos erros: os residuos não estão correlacionados entre si.

 - Especificação correta: o modelo inclui todas as variáveis relavante e não há omissão de termos importantes.

Se alguma dessas condições for violada, então os erros padrão calculados podem estar incorretos - e, consequentemente, os valores-t e p-valores podem não ser confiáceis para testar hipóteses sobre os coeficientes.

```python
# Extraindo os coeficientes do modelo (o que o modelo aprendeu durante o treinamento)
b0 = modelo.params['const']
b1 = modelo.params['Horas_Estudo']
```

```python
# Cria o gráfico
sns.regplot(data = df_score,
            x = 'Horas_Estudo',
            y = 'Nota',
            ci = None,
            line_kws = {'color': 'red'})
# Título e rótulos
plt.title('Regressão Linear: Horas de Estudo x Nota no Exame')
plt.xlabel('Horas de Estudo')
plt.ylabel('Nota no Exame')

# Adiciona a formula e o R² no gráfico
plt.text(x = df_score['Horas_Estudo'].min(),
         y = df_score['Nota'].max() - 2,
         s = f'y = {b0:.2f} + {b1:.2f}x\nR² = {modelo.rsquared:.2f}',
         fontsize = 12,
         color = 'black',
         bbox = dict(facecolor = 'yellow', alpha = 0.7, edgecolor = 'gray'))
plt.show()
```

**Nota: Todo o modelo é um aproximação e isso é suficiente para fazer inferências e análises (de acordo com a performance do modelo), sempre sabendo que há uma margem de erro. Isso vale para qualquer modelo de inteligência artificial na terra.**
